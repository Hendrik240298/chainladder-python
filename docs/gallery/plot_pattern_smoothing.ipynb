```python
def calc_r_2(dev_exact, dev_pred):
    y_exact = dev_exact.ldf_['incurred'].values[0,0,0,:]
    y_pred = dev_pred.ldf_['incurred'].values[0,0,0,:y_exact.size]
    
    weights = np.ones_like(y_exact)
    # weights = y_exact/np.max(y_exact)
    # print(f"weights: {weights}")
    w_sum = np.nansum(weights, axis=0)
    y_mean = np.nansum(weights * y_exact) / w_sum
    
    tss = np.nansum(weights * (y_exact - y_mean) ** 2)
    rss = np.nansum(weights * (y_exact - y_pred) ** 2)

    return 1 - rss/tss
```

```python
print(calc_r_2(dev_smooth, tail_inverse))
print(calc_r_2(dev_smooth, tail_weibull))
```
