```python
import chainladder as cl 
import pandas as pd
import numpy as np
```

```python
quarterly = cl.load_sample('quarterly')
quarterly.link_ratio
```

```python
dev = cl.Development().fit_transform(quarterly)
tail = cl.TailCurve().fit(dev)
tail.ldf_["incurred"]
```

```python
dev.link_ratio['incurred'].heatmap()
```

```python
dev_smooth = cl.Development(
    # drop=[
    #     ('2001',12), ('2001',15), #('2001',6),
    #     ('2002', 21),
    #     ('2003',9), ('2003', 12), ('2003', 18), ('2003', 24)
    # ],
).fit_transform(quarterly)

tail_smooth = cl.TailCurve(
    curve='weibull',
    fit_period=(3, 15),
    attachment_age=6

    ).fit_transform(dev_smooth)
tail_smooth.link_ratio['incurred'].heatmap()
#

# 'incurred' is in the columns dimension (vdims), not the index (kdims)
incurred_idx = list(dev_smooth.ldf_.vdims).index('incurred')
dev_smooth.ldf_.values[:, incurred_idx:incurred_idx+1, :, :4] = tail_smooth.ldf_.values[:, incurred_idx:incurred_idx+1, :, :4]



tail_smooth.ldf_['incurred']
```

```python
dev_smooth.ldf_['incurred']
```

```python
def calc_r_2(dev_exact, dev_pred):
    y_exact = dev_exact.ldf_['incurred'].values[0,0,0,:]
    y_pred = dev_pred.ldf_['incurred'].values[0,0,0,:y_exact.size]
    
    weights = np.ones_like(y_exact)
    
    w_sum = np.nansum(weights, axis=0)
    y_mean = np.nansum(weights * y_exact) / w_sum
    
    tss = np.nansum(weights * (y_exact - y_mean) ** 2)
    rss = np.nansum(weights * (y_exact - y_pred) ** 2)

    return 1 - rss/tss
```

```python
print(calc_r_2(dev_smooth, tail_smooth))
print(calc_r_2(dev, tail_smooth))
```
